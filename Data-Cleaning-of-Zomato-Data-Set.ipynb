### Task 2: clean column names

```python
# strip() = removes spaces at both the start and the end
# .lower = makes letter small 
# .replace = replaces two string 
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

### Task 3: remove duplicate rows

```python
print("rows after :", len(df))
```

### Task 4a: fix the name column

```python
print(df["name"].head(15))
# before
```

```python
df["name"] = df["name"].str.strip().str.title()
# strip() removes extra spaces
# title() makes "BLUE BOWL" -> "Blue Bowl"
```

```python
print(df["name"].head(15))
```

### Task 4b: fix the city column

```python
print(df["city"].head(15))
```

```python
df["city"] = df["city"].str.strip().str.title()
# strip() removes extra spaces
# title() makes "BLUE BOWL" -> "Blue Bowl"
```

```python
df["city"] = df["city"].replace({"Bangalore": "Bengaluru", "New Delhi": "Delhi"})
# Replace old names with new names 
# like Bangalore -----> Bengaluru  AND New Delhi -----> Delhi
```

```python
print(df["city"].value_counts())
```

### Task 4c: fix the cuisine column

```python
print(df["cuisine"].head(15))
```

```python
df["cuisine"] = df["cuisine"].str.strip().str.title()
# strip() removes extra spaces from both sides 
# title() makes "BLUE BOWL" -> "Blue Bowl"
```

```python
print(df["cuisine"].isna().sum())
# there are 12 rows of cuisines are missing 
```

```python
df["cuisine"] = df["cuisine"].fillna("Unknown")       # fill that 12 rows with word "Unknown"
```

```python
print(df["cuisine"].isna().sum())
# NOW there are 0 rows of cuisines are missing 
```

### Task 5: fix cost_for_two

```python
df["cost_for_two"] = (df["cost_for_two"]                      # choose columns
                      .str.replace("₹", "", regex=False)      # replace Rupee Sign with ""
                      .str.replace("Rs", "", regex=False)     # replace "Rs" with ""
                      .str.replace(",", "", regex=False)      # replace "," with ""
                      .str.strip())                           # strip() removes extra spaces from both sides

# remove "₹", "Rs" and ","
# (regex=False means: treat it as plain text)
```

```python
print(df["cost_for_two"].value_counts())
```

```python
# text -> number. errors="coerce" turns "Not Available" and "-" into NaN
df["cost_for_two"] = pd.to_numeric(df["cost_for_two"], errors="coerce")
```

```python
print(df["cost_for_two"].value_counts())
```

```python
# wrong values. cost 0 or negative is impossible, above 10000 is a typing mistake
bad = (df["cost_for_two"] <= 0) | (df["cost_for_two"] > 10000)
```

```python
print("wrong values:", bad.sum())
```

```python
print(df[bad])
# Those bad values which are 0 and more than 10K
```

```python
df.loc[bad, "cost_for_two"] = np.nan
# It sets the cost_for_two value to NaN (missing) for only the rows where bad is True,
# which are the 5 wrong costs (0 or above 10000).
```

```python
# fill all missing costs with the median cost
```

```python
print(df["cost_for_two"].isna().sum())
# missing values before fill
```

```python
print(df["cost_for_two"].isna().head(17))
```

```python
median_cost = df["cost_for_two"].median()
print(median_cost)
```

```python
df["cost_for_two"] = df["cost_for_two"].fillna(median_cost)
# fill the nan values with median_cost
```

```python
print(df["cost_for_two"].isna().sum())
# After filling all missing values 
# Remaining nan values in cost_for_two column
```

```python
print(df["cost_for_two"].describe())
```

### Task 6: fix rating

```python
# "NEW" means a brand new restaurant (no rating yet). 
# Remember it in a new True/False column
```

```python
df["is_new"] = df["rating"] == "NEW"
print("new restaurants:", df["is_new"].sum())
# There are 14 NEW restaurants
```

```python
print(df[df["is_new"]])
# Those 14 rows
```

```python
# text -> number ("NEW" becomes NaN because of errors="coerce")
# The rating column was text (like "4.5" and "NEW"), and this line turns it into numbers.
# "4.5" becomes the number 4.5.
# "NEW" can’t be a number, so errors="coerce" turns it into NaN (missing) instead of giving an error.

df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
```

```python
print(df["rating"].dtype)
# NOW rating col is purely float type
```

```python
print(df["rating"].isna().sum())
# NOW there 26 rows which have nan value
```

```python
# a rating above 5 is impossible (8.8, 9.1 ...) -> make it missing
print("rating above 5 :", (df["rating"] > 5).sum())
```

```python
print(df[df["rating"] > 5])
```

```python
df.loc[df["rating"] > 5, "rating"] = np.nan
# It sets rating to NaN (missing) in the rows where the rating is above 5, 
# which are the 3 wrong ratings (7.5, 8.8, 9.1)
```

```python
print("missing after :", df["rating"].isna().sum())
# only the new restaurants are 3 and 26 old restaurents having nan value
```

```python
print(df["rating"].describe())
```

### Task 7: fix votes

```python
# Votes is the number of customers who gave a rating to that restaurant. 
# A restaurant with votes = 800 and rating = 4.5 means 800 people rated it and the average came out as 4.5.
# More votes means the rating is more reliable. A 4.9 rating from 20 votes is less trustworthy than a 4.5 from 2000 votes.
# That’s why a negative vote count is impossible, and why we made those values missing.
```

```python
print(df["votes"].sum())
# Total votes taken by people
```

```python
print(df["votes"].isna().sum())
# Total missing votes
```

```python
print(df[df["votes"].isna()])
# That 10 missing votes before cleaning
```

```python
print(df["votes"].min(), df["votes"].max())   
# range of votes
```

```python
print(df["votes"].nunique())                  
# how many different votes made by people
```

```python
print(df["votes"].unique())                   
# the different votes made by people
```

```python
print(df["votes"].dtype)
```

```python
print("Positive and Negative votes :\n", (df["votes"] < 0).value_counts())
```

```python
print("negative votes:", (df["votes"] < 0).sum())
# Total negative votes
```

```python
print(df[df["votes"] < 0])
```

```python
df.loc[df["votes"] < 0, "votes"] = np.nan
# Converting negative votes into nan --------> 4 (-ve) + 10 (nan) = 14 (nan)
```

```python
print(df["votes"].isna().sum())
# Total nan 
```

```python
df["votes"] = df["votes"].fillna(df["votes"].median())
# Fill df["votes"] column with median of votes
```

```python
print(df["votes"].isna().sum())
# Total missing votes (nan)
```

```python
print(df["votes"].dtype)
# current data type
```

```python
df["votes"] = df["votes"].astype(int)
```

```python
print(df["votes"].dtype)
# Data type after changing to int
```

```python
print(df["votes"].describe())
```

### Task 8: fix online_order (Yes/yes/Y/No/N ... -> True/False)

```python
print(df["online_order"].value_counts())
```

```python
df["online_order"] = df["online_order"].str.strip().str.lower()
# .strip() removes extra spaces
# .lower() makes that row string into small letters
```

```python
print(df["online_order"].value_counts())
```

```python
df["online_order"] = df["online_order"].map({"yes": "yes", "y": "yes", "no": "no", "n": "no"})
# map every spelling to just "yes" or "no"
```

```python
print(df["online_order"].value_counts())
```

```python
print("missing:", df["online_order"].isna().sum())
# Total Missing values 
```

```python
print("most common:", df["online_order"].mode()[0])
# Most common answer for online_order is no
```

```python
df["online_order"] = df["online_order"].fillna(df["online_order"].mode()[0])
# So fill missing values by most common answers that is No
```

```python
print("missing:", df["online_order"].isna().sum())
```

```python
print(df["online_order"].value_counts())
```

```python
df["online_order"] = df["online_order"] == "yes"
# "yes", "no", "yes" becomes True, False, True
```

```python
print(df["online_order"].value_counts())
```

```python
print(df["online_order"].dtype)
```

### Task 9: fix opening_date and add opening_year

```python
print(df["opening_date"].dtype)
```

```python
print(df["opening_date"].head(10))
```

```python
print("missing dates:", df["opening_date"].isna().sum())
```

```python
print(df["opening_date"].value_counts().sum())
# Total different values in opening_date column
```

```python
df["opening_date"] = pd.to_datetime(df["opening_date"], format="%Y-%m-%d", errors="coerce")
# Convert text -> real date --------> pd.to_datetime 
# format="%Y-%m-%d" eg. 2012-10-17  -----> year-month-date
# "Unknown" and empty cells become NaT (= missing date)
```

```python
print(df["opening_date"].value_counts().sum())
```

```python
print("missing dates:", df["opening_date"].isna().sum())
# 10 Missing date and 5 Unknown dates
```

```python
print((df["opening_date"] >= "2026-10-30").sum())
# a restaurant cannot open in the future (today is 30 Oct 2026 ) -> make those missing
```

```python
print(df[df["opening_date"] >= "2026-10-30"])
```

```python
future = df["opening_date"] >= "2026-10-30"
```

```python
print(future.sum())
```

```python
df.loc[future, "opening_date"] = pd.NaT
# for the rows where future is True , It sets opening_date to NaT (missing date)
```

```python
print((df["opening_date"] >= "2026-10-30").sum())
```

```python
print(df[df["opening_date"] >= "2026-10-30"])
```

#### We don’t fill the missing dates, because a made-up date would be wrong data.

```python
df["opening_year"] = df["opening_date"].dt.year
# makes a new column holding only the year from each date, for example 2014-08-09 becomes 2014. 
# A missing date gives a missing year.
```

```python
print("missing dates now:", df["opening_date"].isna().sum())
# counts the missing dates. It should be 18: 10 blank, 5 “Unknown” and 3 future.
```

```python
print(df[["opening_date", "opening_year"]].head())
# shows the date and year side by side for the first 5 rows, so you can see the year was taken correctly.
```

```python
print("Min value of date :",df["opening_date"].min(),"AND Max value of date :", df["opening_date"].max())
# shows the oldest and newest opening date. This confirms no impossible dates are left.
```

---
#### Text to dates. pd.to_datetime(..., errors="coerce") turned the text into real dates. “Unknown” and blanks became NaT (15 missing).
#### Future dates. The 3 dates in 2027 were set to NaT with df.loc[future, "opening_date"] = pd.NaT. Now 18 are missing.
#### New column. opening_year was created with .dt.year.
---

### Task 10: make 2 new columns

#### pd.cut puts numbers into groups:  `(0,300] cheap | (300,700] medium | (700,10000] costly`

```python
df["cost_level"] = pd.cut(df["cost_for_two"], bins=[0, 300, 700, 10000], labels=["Cheap", "Medium", "Costly"])
```

```python
print(df[["cost_for_two", "cost_level"]].head(10))
# New DataFrame Column
```

```python
print(df["cost_level"].value_counts())
# Total Costly , Medium , cheap -----> Resaurents based on spend of customers
```

#### you can read the quality at a glance instead of reading the number

```python
df["rating_label"] = np.where(df["rating"] >= 4, "Good", "Average")
# np.where(condition, value_if_true, value_if_false)
```

```python
df.loc[df["is_new"], "rating_label"] = "New"
# note: new restaurants have no rating, so make their label "New"
```

```python
print(df[["rating", "rating_label"]].head(10))
```

```python
print(df["rating_label"].value_counts())
# Total Good , Average and New restaurents --------> based on Rating from Customers
```
